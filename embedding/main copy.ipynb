```python
import dotenv
import os
from google import genai
from google.genai import types
```

```python
GOOGLE_API_KEY = dotenv.get_key('../.env', 'GOOGLE_API_KEY')
DOCUMENT_URL = '../crawler/output/'
```

```python
files : list[str] = os.listdir(DOCUMENT_URL)
docs : list[str] = []
for file_name in files:
    with open(DOCUMENT_URL+file_name, encoding="utf-8") as f:
        text = f.read()
    
    docs.append(text)

for doc in docs :
    print(doc)
    print()
    print()
    print()
    print("=======================================")
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("BAAI/bge-base-en-v1.5", device="cuda")
embeddings = model.encode(docs, batch_size=32, show_progress_bar=True)
```

```python
print(len(embeddings), len(embeddings[0]))
for e in embeddings : 
    print(e)
```

```python
import pickle
with open("embeddings.pkl", "wb") as f:
    pickle.dump(embeddings, f)
```
